## 1. ORGANIZACIÓN DEL ESTABLECIMIENTO

### Entidades:
### 1.1 Campo

**Atributos principales**
- idCampo
- nombre
- ubicación (localidad)
- superficie

### 1.2 Lote

**Atributos principales**

- idLote
- nombre (1,2,...)
- superficie

### 1.3 Sector

**Atributos principales**

- idSector
- nombre (A, B, C...)

### 1.4 Pivote

**Atributos principales**

- idPivote
- nombre (1, 2,...)
- fabricante
- modelo

### 1.5 AsignaciónPivote

Representa el historial de asignación de un pivote a un lote.

**Atributos principales**

- idAsignacion
- fechaInicio
- fechaFin

### Relaciones:
- Un **Campo** posee múltiples **Lotes**.
- Un **Campo** posee múltiples **Pivotes**.
- Un **Lote** pertenece a un único **Campo**.
- Un **Lote** puede dividirse en múltiples **Sectores**.
- Un **Sector** pertenece a un único **Lote**.
- Un **Pivote** puede ser asignado a distintos **Lotes** a lo largo del tiempo mediante la entidad **Asignación de Pivote**. 
- Un **Lote** puede tener asignado distintos **Pivotes** a lo largo del tiempo mediante la entidad **Asignación de Pivote** 

### Cardinalidades

| Relación | Cardinalidad |
|----------|--------------|
| Campo — Lote | 1 : N |
| Campo — Pivote | 1 : N |
| Lote — Sector | 1 : N |
| Pivote — Asignación de Pivote | 1 : N |
| Lote — Asignación de Pivote | 1 : N |

### Restricciones del dominio

- Un lote pertenece a un único campo.
- Un sector pertenece a un único lote.
- Un pivote pertenece a un único campo.
- Un pivote puede regar distintos lotes a lo largo del tiempo, uno a la vez.
- Un lote puede ser regado por distintos pivotes a lo largo del tiempo, uno a la vez.

## 2. Dispositivos IoT

### Entidades:

### 2.1 Categoría de Dispositivo

**Atributos principales**

- idCategoria
- nombre (suelo, riego, meteorológico)


### 2.2 Tipo de Dispositivo

**Atributos principales**

- idTipo
- nombre (sonda, ph, pluviometro, presion, caudal, wts)

### 2.3 Variable

**Atributos principales**

- idVariable
- nombre
- unidad

### 2.4 Dispositivo

**Atributos principales**

- idDispositivo
- fabricante
- modelo
- númeroSerie
- dev_eui
- app_eui
- app_key
- at_pin
- ota_pin
- intervaloTransmision (intervalo de tiempo en segundos)
- estadoOperativo (on / off)
- estadoComunicacion (connected / disconnected)


### 2.5 InstalaciónDispositivo

Representa el historial de ubicación de un dispositivo dentro del establecimiento.

**Atributos principales**

- idInstalacion
- fechaInicio
- fechaFin

### Relaciones

- Una **Categoría de Dispositivo** contiene múltiples **Tipos de Dispositivo**.
- Un **Tipo de Dispositivo** clasifica múltiples **Dispositivos**.
- Un **Tipo de Dispositivo** sensa múltiples **Variables**.
- Un **Dispositivo** puede tener múltiples **Instalaciones** a lo largo de su vida útil.
- Cada **Instalación** corresponde a un único **Dispositivo**.
- Una **Instalación** puede asociarse a un **Campo**, un **Sector** o un **Pivote**.
- Un **Campo** puede tener múltiples **Instalaciones**.
- Un **Sector** puede tener múltiples **Instalaciones**.
- Un **Pivote** puede tener múltiples **Instalaciones**.

### Cardinalidades

| Relación | Cardinalidad |
|----------|--------------|
| Categoría de Dispositivo — Tipo de Dispositivo | 1 : N |
| Tipo de Dispositivo — Dispositivo | 1 : N |
| Tipo de Dispositivo — Variable | 1 : 1..N |
| Dispositivo — Instalación | 1 : N |
| Campo — Instalación | 1 : N |
| Sector — Instalación | 1 : N |
| Pivote — Instalación | 1 : N |

### Restricciones del dominio
- Un dispositivo pertenece a un único tipo de dispositivo.
- Un tipo de dispositivo pertenece a una única categoría.
- Un tipo de dispositivo puede sensar una o varias variables.
- Todos los dispositivos sensan la variable "batería".
- Una variable es sensada por un único tipo de dispositivo.
- Un dispositivo puede tener cero a múltiples instalaciones a lo largo del tiempo, pero solo una puede estar activa simultáneamente.
- Cada instalación debe estar asociada a un único Campo, un único Sector o un único Pivote.
- Un Campo, Sector o Pivote puede tener múltiples dispositivos instalados.
- Un dispositivo sin una instalación activa no puede encontrarse encendido.
- Un dispositivo apagado no puede encontrarse conectado a la red de comunicaciones.
- Un dispositivo conectado debe encontrarse encendido y poseer una instalación activa.

## 3. Mediciones
Representa cada medición realizada por un dispositivo. Los valores medidos se almacenan en formato JSONB, permitiendo registrar una o varias variables según el tipo de dispositivo.

### Entidades

### 3.1 Gateway

**Atributos principales**

- idGateway
- nombre


### 3.2 Medición

Representa cada medición realizada por un dispositivo.

**Atributos principales**

- idMedicion
- fechaHora
- valoresMedidos (JSONB)
- RSSI
- SNR
- contadorMensajes (numeración ascendente)

### Relaciones
- Un **Dispositivo** genera múltiples **Mediciones**.
- Cada **Medición** pertenece a un único **Dispositivo**.
- Una **Medición** tiene múltiples **Variables**.
- Un **Gateway** recibe múltiples **Mediciones**.

### Cardinalidades

| Relación | Cardinalidad |
|----------|--------------|
| Dispositivo — Medición | 1 : N |
| Medición — Variable | 1 : N |
| Gateway — Medición | 1 : N |

### Restricciones del dominio
- Cada medición pertenece a un único dispositivo.
- Cada medición es recibida por un único gateway.
- No existen campos con datos nulos en una medición.

## 4. Alarmas

### Entidades
### 4.1 ReglaAlarma

Define las condiciones bajo las cuales el sistema debe generar una alarma. Acutalmente se contemplan dos posibilidades: batería baja o alguna/s variable/s fuera de rango.

**Atributos principales**

- idRegla
- descripcion (ej: Alta presión en tubería.. Estrés ídrico..batería baja.. )
- umbralInferior
- umbralSuperior
- habilitada


### 4.2 EventoAlarma

Representa el evento generado cuando una regla de alarma se cumple.

**Atributos principales**

- idEventoAlarma
- fechaHora
- valorDetectado

### Relaciones
- Un **EventoAlarma** se genera cuando una **Medición** cumple una **ReglaAlarma**.
- Una **ReglaAlarma** puede aplicarse a uno o varios **Dispositivos**.
- Un **Dispositivo** puede tener asociadas una o varias **ReglaAlarma**.
- Una **Variable** puede estar asociada a múltiples **ReglaAlarma**.
- Una **ReglaAlarma** evalúa una única **Variable**.
- Una **ReglaAlarma** puede generar múltiples **EventoAlarma** (del mismo dispositivo a lo largo del tiempo o de distintos dispositivos).
- Cada **EventoAlarma** es generado por una única **ReglaAlarma**.
- Una **Medición** puede generar ninguno, uno o varios **EventoAlarma**.

### Cardinalidades
| Relación | Cardinalidad |
|----------|--------------|
| ReglaAlarma — Dispositivo | N : M |
| ReglaAlarma — EventoAlarma | 1 : N |
| Variable — ReglaAlarma | 1 : N |
| ReglaAlarma — Variable | 1 : 1 |
| Medición — EventoAlarma | 1 : 0..N |

### Restricciones del dominio

- Una regla de alarma define la condición para generar una alarma.
- Una regla de alarma puede establecer un umbral inferior, un umbral superior o ambos, según el comportamiento que se desee monitorear.
- Una regla de alarma debe establecer como mínimo un umbral.
- La variable asociada a una regla debe pertenecer al tipo de dispositivo sobre el cual se aplica la regla.

## 5. Usuarios

### Entidades
### 5.1 Perfil

**Atributos principales**

- idPerfil
- nombre
- descripción


### 5.2 Usuario

**Atributos principales**

- idUsuario
- nombre
- apellido
- email
- contraseña

### Relaciones
- Un **Perfil** puede estar asociado a múltiples **Usuarios**.
- Un **Usuario** pertenece a un único **Perfil**.

### Cardinalidades
| Relación | Cardinalidad |
|----------|--------------|
| Perfil — Usuario | 1 : N |

### Restricciones del dominio
- Existen sólo 3 perfiles: Operador, Configurador y Administrador
- Todos los usuarios deben tener asignado un perfil.
- Ningun usuario puede tener mas de un perfil.
- Un usuario no puede tener datos nulos.
